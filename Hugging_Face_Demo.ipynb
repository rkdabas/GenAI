```python
# working of above code
from transformers import AutoTokenizer

# load a pre-trained tokenizer
tokenizer=AutoTokenizer.from_pretrained("bert-base-uncased")

# example text
text="I am feeling great"

# tokenize the text
tokens=tokenizer.tokenize(text)
print("tokens: ",tokens)

# convert tokens to input ids (vector representation)
input_ids=tokenizer.convert_tokens_to_ids(tokens)
print("input ids: ",input_ids)

# encode the text (tokenization + converting to input ids)
encoded_input=tokenizer(text)
print("encoded input: ",encoded_input)

# decode the text
decoded_text=tokenizer.decode(input_ids)
print("decoded output: ",decoded_text)
```

Fine Tuning IMDB Dataset

```python
# step1: using hugging face dataset
!pip install datasets
```

```python
#  step 2: load and prepare the dateset
# imdb is name of dataset in hugging face
from datasets import load_dataset
dataset=load_dataset('imdb')
```

```python
dataset
```

```python
dataset['train'][0]
```

```python
# step3:  pre-process the data

# tokenize the dataset using the tokenizer associated with the pre-trained model
from transformers import AutoTokenizer

# load the tokenizer
tokenizer=AutoTokenizer.from_pretrained("bert-base-uncased")

# tokenizd the dataset
def tokenize_function(examples):
  return tokenizer(examples["text"],padding="max_length",truncation=True)

tokenized_dataset=dataset.map(tokenize_function,batched=True)
```

```python
tokenized_dataset
```

```python
tokenized_dataset['train'][0]
```

```python
# step4: set up the training arguments for hyper-parameter tuning

# specify the hyper-parameter and training settings

from transformers import TrainingArguments
training_args=TrainingArguments(
    output_dir='./results',
    eval_strategy='epoch',
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=1,
    weight_decay=0.01,
)
training_args
```

```python
# step5 : initialize the model

# load the pre-trained model and define the training procodure
# num_labels=2 because we have only 2 sentiments here i.e positive and negative
from transformers import AutoModelForSequenceClassification, Trainer

# load the pre-trained model
model=AutoModelForSequenceClassification.from_pretrained('bert-base-uncased',num_labels=2)

# initialize the trainer
trainer=Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset['train'],
    eval_dataset=tokenized_dataset['test']
)
```

```python
# step 6: train the model
trainer.train()
```

```python
# step 7: evaluate the mode

# assess the model's performance on a validation set
results=trainer.evaluate()
print(results)
```

```python
# step 8 : save the fine-tuned model and tokenizer
model.save_pretrained('./fine-tuned-model')
# if user gives more data in future which needs to be tokenized too. So save the tokenizer for future tokenization
tokenizer.save_pretrained('./fine-tuned-model')
```

```python
import arxiv
import pandas as pd
```

```python
# query to fetch the AI-related papers
query='ai or Artifical intelligence or machine learning'
search=arxiv.Search(query=query, max_results=10, sort_by=arxiv.SortCriterion.SubmittedDate)


# fetch the papers
papers=[]
for result in search.results():
  papers.append({
      'published':result.published,
      'title':result.title,
      'abstract':result.summary,
      'categories':result.categories
  })

  # convert to dataframe
  df=pd.DataFrame(papers)

  pd.set_option('display.max_colwidth',None)
```

```python
  df.head(10)
```

```python
# summarizing the abstract column
from transformers import pipeline

abstract=df['abstract'][0]
summarizer=pipeline("summarization",model="facebook/bart-large-cnn")

# summarization
summarization_result=summarizer(abstract)
```
